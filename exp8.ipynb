```python

# K-Nearest Neighbour without libraries

# Read dataset
file = open("iris.csv", "r")
file.readline()
data = []

for line in file:
    v = line.strip().split(",")
    if len(v) == 5:
        data.append([
            [float(v[0]), float(v[1]),
             float(v[2]), float(v[3])],
            v[4]
        ])

file.close()

# Calculate Euclidean distance
def distance(a, b):
    total = 0
    for i in range(4):
        total += (a[i] - b[i]) ** 2
    return total ** 0.5

# Predict using KNN
def predict(train, point, k):
    distances = []

    for features, species in train:
        distances.append([distance(features, point), species])

    distances.sort()
    votes = {}

    for d, species in distances[:k]:
        votes[species] = votes.get(species, 0) + 1

    return max(votes, key=votes.get)

# Split dataset
train = data[:120]
test = data[120:]

# Calculate accuracy
k = 5
correct = 0

for features, actual in test:
    if predict(train, features, k) == actual:
        correct += 1

print("Accuracy:", correct / len(test) * 100, "%")

# Get user input
point = [
    float(input("Sepal Length: ")),
    float(input("Sepal Width: ")),
    float(input("Petal Length: ")),
    float(input("Petal Width: "))
]

print("Predicted Species:", predict(train, point, k))
```
